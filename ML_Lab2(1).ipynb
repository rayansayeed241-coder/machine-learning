```python
#Code for outlier detection and removing outlier with IQR method

Q1 = df["Price"].quantile(0.25)
Q3 = df["Price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

df = df[(df["Price"] >= lower) & (df["Price"] <= upper)]
```

# Single Linear Regression.

```python
x = df[['Avg. Area Income']]
y = df['Price']
```

```python
# 90:10 is high risk split for training and testing
# 40:60 is not valid for training and testing
# 85:15 is valid for training and testing
# Accuracy is R2 score

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)

```

```python
from sklearn.linear_model import LinearRegression

slr_model = LinearRegression()
slr_model.fit(x_train, y_train)
```

```python
y_pred = slr_model.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import numpy as np

print("Simple Linear Regression Results")
print("--------------------------------")

print("MAE :",mean_absolute_error(y_test, y_pred))
print("MSE :",mean_squared_error(y_test, y_pred))
print("RMSE:",np.sqrt(mean_squared_error(y_test, y_pred)))
print("R2 Score:", r2_score(y_test, y_pred))
```

```python
print("Intercept:", slr_model.intercept_)
print("Slope:", slr_model.coef_[0])
```

```python
plt.figure(figsize=(8,6))
plt.scatter(x_test,y_test,color='blue',label='Actual Data')
plt.plot(x_test,y_pred,color='red', linewidth=2,label='Regression Line')

plt.xlabel("Avg. Area Income")
plt.ylabel("House Price")
plt.title("Simple Linear Regression")
plt.legend()

plt.show()
```

```python
income = float(input("Enter Average Area Income:"))
new_data = np.array([[income]])
prediction = slr_model.predict(new_data)
print("Predicted House Price = ${:,.2f}".format(prediction[0]))
```

```python
# For interview hyperparameter is imp.
# Hyperparameter applies only on training dataset.
# We use hyperparameter to overcome the problem of overfit or to increase the accuracy of model.

# overfit -> L2(ridge) regulisation -> reduce coeff to 0.
# when more outliers -> L1(lasso) regulisation -> makes coeff directly 0.
# when both overfit and more outliers -> L1 + L2(elastic net)
```

# Multiple Linear Reression.

```python
x = df[['Avg. Area Income',
     'Avg. Area House Age',
     'Avg. Area Number of Rooms',
     'Avg. Area Number of Bedrooms',
     'Area Population']]
y = df['Price']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
mlr_model = LinearRegression()
mlr_model.fit(x_train, y_train)
```

```python
y_pred = mlr_model.predict(x_test)
```

```python
print("Multiple Linear Regression Results")
print("--------------------------------")

print("MAE :",mean_absolute_error(y_test, y_pred))
print("MSE :",mean_squared_error(y_test, y_pred))
print("RMSE:",np.sqrt(mean_squared_error(y_test, y_pred)))
print("R2 Score:", r2_score(y_test, y_pred))
```

```python
print("Intercept:", mlr_model.intercept_)
print("Slope:", mlr_model.coef_[0])
```

```python
income = float(input("Enter Average Area Income: "))
house_age = float(input("Enter Average Area House Age: "))
rooms = float(input("Enter Average Area Number of Rooms: "))
bedrooms = float(input("Enter Average Area Number of Bedrooms: "))
population = float(input("Enter Area Population: "))

new_data = [[income, house_age, rooms, bedrooms, population]]

predicted_price = mlr_model.predict(new_data)

print("\nPredicted House Price =", predicted_price[0])
```

# Ridge Regression

```python
from sklearn.model_selection import GridSearchCV
```

```python
from sklearn.linear_model import Ridge
ridge = Ridge()
ridge.fit(x_train,y_train)
ridge_pred = ridge.predict(x_test)
print("Default Ridge R2 :", r2_score(y_test, ridge_pred))
```

```python
param_grid = {'alpha':[0.001,0.01,0.1,1,10,100]}
```

```python
grid_ridge = GridSearchCV(estimator=Ridge(),param_grid=param_grid,scoring='r2',cv=5)
```

```python
grid_ridge.fit(x_train, y_train)
```

```python
print("Best Alpha :",grid_ridge.best_params_)
```

```python
print(grid_ridge.best_score_)
```

```python
best_ridge = grid_ridge.best_estimator_
ridge_pred = best_ridge.predict(x_test)
```

```python
print("MAE :",mean_absolute_error(y_test, ridge_pred))
print("MSE :",mean_squared_error(y_test, ridge_pred))
print("RMSE:",np.sqrt(mean_squared_error(y_test, ridge_pred)))
print("R2 Score:", r2_score(y_test, ridge_pred))
```
